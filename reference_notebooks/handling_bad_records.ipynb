# Handling Bad Records in Dataframes

## Reference Block for masking & marking bad records in Dataframes

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {'A': [1, 2, 3, 4, 5],
        'B': [6, 7, 8, 9, 10],
        'C': [11, 12, 13, 14, 15]}
df = pd.DataFrame(data)

# Create a mask (boolean array)
mask = df > 3

# Apply the mask to replace values greater than 3 with -1
df_masked = df.mask(mask, -1)

print("Original DataFrame:")
print(df)
print("\nMasked DataFrame:")
print(df_masked)

# Using a callable function in mask
df_masked_callable = df.mask(lambda x: x % 2 == 0, 'even')
print("\nMasked DataFrame with callable:")
print(df_masked_callable)

# Using mask for NaN replacement
df_with_nan = df.copy()
df_with_nan.iloc[1, 1] = np.nan
print("\nDataFrame with NaN:")
print(df_with_nan)

df_nan_filled = df_with_nan.mask(df_with_nan.isna(), 0)
print("\nDataFrame with NaN filled:")
print(df_nan_filled)
```

## Reference Block for masking & dropping bad records in Dataframes

```python
import pandas as pd
import numpy as np

# Sample DataFrame with NaN values
data = {'A': [1, 2, np.nan], 'B': [5, np.nan, np.nan], 'C': [1, 2, 3]}
df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# Drop rows with any NaN values
df_dropna_rows = df.dropna()
print("\nDataFrame after dropping rows with NaN:\n", df_dropna_rows)

# Drop columns with any NaN values
df_dropna_cols = df.dropna(axis=1)
print("\nDataFrame after dropping columns with NaN:\n", df_dropna_cols)

# Drop rows where all values are NaN
df_dropna_all_rows = df.dropna(how='all')
print("\nDataFrame after dropping rows with all NaN:\n", df_dropna_all_rows)

# Drop columns where all values are NaN
df_dropna_all_cols = df.dropna(axis=1, how='all')
print("\nDataFrame after dropping columns with all NaN:\n", df_dropna_all_cols)

# Drop rows with NaN in specific columns
df_dropna_subset = df.dropna(subset=['A', 'B'])
print("\nDataFrame after dropping rows with NaN in columns A or B:\n", df_dropna_subset)

# Modify the original DataFrame
df_copy = df.copy()
df_copy.dropna(inplace=True)
print("\nDataFrame after dropping NaN in place:\n", df_copy)
```
