## Plotting with Functions

```python
# Here I have written a custom Python function that can be used to create the same plot
def deg2rad(theta):   # note this functions accepts degrees and returns radians
    return theta * np.pi / 180 # convert degrees to radians
```

```python
# AN EXAMPLE OF A SCATTER PLOT OF TWO INDEPENDENT VARIABLES 'y' vs 'x'
# CREATING CIRCLE SCATTER PLOT USING THE PARAMETRIC EQUATION OF THE CIRCLE

def yval(angle):
    return np.sin(angle)

def xval(angle):
    return np.cos(angle)

theta = np.linspace(0,2*np.pi, num=50) ## the linspace method defines the actual number angles within the range

plt.title("A parametric Circle of scattered XY points")
plt.xlabel("xval")
plt.ylabel("yval");
plt.gca().set_aspect('equal')
# Note the use of scatter method
plt.scatter(xval(theta),yval(theta))
plt.show()
```

```python
# AN EXAMPLE OF A SCATTER PLOT OF TWO INDEPENDENT VARIABLES 'y' vs 'x'
# CREATING CIRCLE PLOT WITH OFFSETS USING THE PARAMETRIC EQUATION OF THE CIRCLE

rad = 5
u = 3
v = -4

theta = np.linspace(0,2*np.pi, num=20) ## the linspace method defines the actual number angles within the range
# create to the 2 independent variables we need for scatter plotting the circle
x = xval(theta)
y = yval(theta)

plt.text(2.8,-3.8,"radius: %1d" % rad)
plt.text(u+0.05,v-0.05,"center: (%1.1f, %1.1f)" % (u,v))
plt.title("An offset Circle with line and dots!")
plt.xlabel("xval")
plt.ylabel("yval");

plt.xlim(0, 4)   ## set x axis range
plt.ylim(-5, 0)   ## set y axis range

plt.gca().set_aspect('equal')

plt.plot(u+xval(theta),v+yval(theta))    # plot line
plt.scatter(u+xval(theta),v+yval(theta)) # plot dots
plt.scatter(u,v,marker='+')  # plot center point # plot center
plt.show()
```

```python
# AN EXAMPLE OF A SCATTER PLOT OF TWO INDEPENDENT VARIABLES 'y' vs 'x'
# CREATING CIRCLE SCATTER & LINE GRID PLOTs USING THE PARAMETRIC EQUATION OF THE CIRCLE

# creating functions yval & xval
def yval(angle):
    return np.sin(angle)

def xval(angle):
    return np.cos(angle)

# create to the 2 independent variables we need for scatter plotting the circle
# create angles theta first
theta = np.linspace(0,2*np.pi, num=20) ## the linspace method defines the actual number angles within the range
# create the 2 parametrized independent variables using functions above
x = xval(theta)
y = yval(theta)

plt.subplot(121)
plt.title("Circle with scatter points!")
plt.xlabel("xval")
plt.ylabel("yval");
plt.gca().set_aspect('equal')
plt.scatter(x,y)

plt.subplot(122)
plt.title("Circle with lines!")
plt.xlabel("xval")
plt.ylabel("yval");
plt.gca().set_aspect('equal')
plt.plot(x,y)

plt.show()
```

```python
##.    PROBLEM #1
a = 3 # semi major axis
b = 2 # semi minor axis
u = 3 # center x coordinate
v = 3 # center y coordinate
c = (a**2-b**2)**0.5 # semi focal distance
e = c/a # eccentricity

# create functions yval & xval - copy & modify from above
def yval(angle):
    return b * np.sin(angle)

def xval(angle):
    return a * np.cos(angle)
    
# create angles 't'
t = np.linspace(0,2*np.pi, num=200)
t1 = np.linspace(3*np.pi/4, np.pi, 10)

# create the 2 parametrized independent variables using your newly created functions above
x = xval(t)
y = yval(t)
x1 = xval(t1)
y1 = yval(t1)

# invoke the plt.plot() method and plt.show() method - copy & modify from above 
plt.title("An Ellipse!")
plt.xlabel("xval")
plt.ylabel("yval");
plt.gca().set_aspect('equal')

# offset x by u and y by v and plot, show again - copy & modify from above
#plt.plot(x,y)
plt.plot(u+x,v+y)
plt.scatter(u+x1,v+y1)

# Add scatter points for center, left focus and right focus of ellipse - copy & modify from above
plt.scatter(u,v,marker='+')
plt.scatter(u-c,v,marker='o')
plt.scatter(u-a,v,marker='o')
#plt.scatter(u+c,v,marker='o')

# Add annotations by using text() method - copy & modify from above
plt.text(u,v+1,"Semi Major Axis: %.1f AU" % a, ha='center', va='center',)
plt.text(u,v+0.5,"Eccentricity: %.2f" % e,  ha='center', va='center',)
```

### More robust examples with multiplots and animations
Matplotlib graphs your data on Figures (e.g., windows, Jupyter widgets, etc.), each of which can contain one or more Axes, an area where points can be specified in terms of x-y coordinates (or theta-r in a polar plot, x-y-z in a 3D plot, etc.). The simplest way of creating a Figure with an Axes is using pyplot.subplots. We can then use Axes.plot to draw some data on the Axes, and show to display the figure:

```python
# Here we will use a more robust way to do plotting
# We are plotting "hard-coded" x and y arrays to plot here
fig, ax = plt.subplots()             # Create a figure containing a single Axes.
ax.plot([1, 2, 3, 4], [1, 4, 2, 3])  # Plot some data on the Axes.
plt.show()                           # Show the figure.
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML, display
plt.style.use('seaborn-v0_8-pastel')

def generate_animation():
    fig, ax = plt.subplots()
    x = np.linspace(0,2*np.pi,1000)
    y = np.sin(x)
    l, = ax.plot(x, y, lw=3)
    
    def animate(i):
        y = np.sin(x+i/(4*np.pi))
        l.set_data([x], [y])
        return [l]

    anim = animation.FuncAnimation(fig, animate, frames=200, interval=20, blit=True)

    display(HTML(anim.to_jshtml()))
    plt.close(fig)

generate_animation()
```
