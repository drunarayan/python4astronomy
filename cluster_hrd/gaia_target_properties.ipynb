### Perform a Google Search about your target

```python
### Setup your target here
target_id = 'betelgeuse'
```

```python
try:
	from googlesearch import search
except ImportError:
	print("No module named 'google' found")

# to search
query = f'{target_id} Wikipedia Astronomy'

for j in search(query, tld="co.in", num=10, stop=10, pause=2):
	print(j)
```

```python
from astroquery.simbad import Simbad
result_table = Simbad.query_objectids(target_id)
        
for x in result_table:
    if 'gaia' in x['ID'].lower():
        print (x['ID'])
result_table
```
